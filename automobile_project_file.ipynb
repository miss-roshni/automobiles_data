## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Load the Dataset

```python
df = pd.read_excel('automobile_data.xlsx')
df.head()
```

## 3. Dataset Information

```python
df.describe()
```

## 4. Print the most expensive car with its price

```python
company_prices=df.groupby('make')['price'].max().sort_values(ascending=False)
```

```python
company_prices.head(1)
```

## 5. Print the total number of cars manufactured by each company

```python
count_make=df['make'].value_counts()
```

# 6.EDA-Total Number of Cars Manufactured by Each Company

```python
plt.figure(figsize=(10,8))
count_make.plot(kind='bar')
plt.title('Total Number of Cars Manufactured by Each Company')
plt.xlabel('Make')
plt.ylabel('Count')
plt.show()
```
